# Visualizing galaxy groups with and without lensing images

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
```

```python
data_true = np.load('data_true.npy')
data_false = np.load('data_false.npy')

print np.shape(data_true), np.shape(data_false)
```

## With Lensing Images

```python
f, axarr = plt.subplots(5, 4, sharex=True, sharey=True, figsize=(15,15), dpi=20)
f.subplots_adjust(hspace=0, wspace=0)

for i in range(5):
    for j in range(4):
        axarr[i,j].imshow(data_true[i*4+j], cmap='Greys')
        axarr[i,j].get_xaxis().set_visible(False)
        axarr[i,j].get_yaxis().set_visible(False)
f.savefig('with.eps')
```

## Without Lensing Images

```python
f, axarr = plt.subplots(5, 4, sharex=True, sharey=True, figsize=(15,15), dpi=20)
f.subplots_adjust(hspace=0, wspace=0)

for i in range(5):
    for j in range(4):
        axarr[i,j].imshow(data_false[i*4+j], cmap='Greys')
        axarr[i,j].get_xaxis().set_visible(False)
        axarr[i,j].get_yaxis().set_visible(False)
        
f.savefig('without.eps')
```
